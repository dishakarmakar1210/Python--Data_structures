1. What are data structures, and why are they important?
-> Data structures are organized ways to store, manage, and retrieve data efficiently, such as arrays, tuples, lists, dictionaries, trees, etc. They are crucial for optimizing algorithms, improving performance, and solving complex computational problems effectively.
2. Explain the difference between mutable and immutable data types with examples?
->Mutable data types can be changed after creation, while immutable data types cannot be modified once assigned.
Mutable: Lists, Dictionaries, Sets
Immutable: Strings, tuples, integers
3. Describe how dictionaries store data?
->Lists and tuples are both used to store collections of data in Python, but they differ in mutability and performance. Lists ([]) are mutable, allowing modifications like adding or removing elements, whereas tuples (()) are immutable and cannot be changed after creation. Lists are ideal for frequently changing data, while tuples are better for fixed or read-only data.
4. Describe how dictionaries store data?
-> Dictionaries in Python store data as key-value pairs using a hash table. Each key is hashed to generate an index. This allows faster look up.
5. Why might you use a set instead of a list in Python?
-> A set is used instead of a list when you need unique elements. Unlike lists, sets do not allow duplicates Use sets for tasks like removing duplicates and set operations like union, intersection, etc.
6. What is a string in Python, and how is it different from a list?
-> A string in Python is a sequence of characters enclosed in quotes ("" or ''), while a list is a collection of elements that can hold different data types.
7. How do tuples ensure data integrity in Python?
-> Tuples ensure data integrity in Python by being immutable, meaning their elements cannot be modified, added, or removed after creation. This prevents accidental changes, making them ideal for storing constant or read-only data.
8. What is a hash table, and how does it relate to dictionaries in Python?
-> A hash table is a data structure that stores key-value pairs using a hash function to map keys to unique indexes for fast retrieval.
In Python, dictionaries (dict) are implemented using a hash table, allowing faster lookups, insertions, and deletions. Each key is hashed to determine its storage location, ensuring efficient access.

14. How does the “in” keyword work differently for lists and dictionaries?
-> In Lists, the 'in' keyword Checks if a value exists in the list (O(n) complexity).
In Dictionaries, the 'in' keyword Checks for the presence of a key (not values) using a hash table (O(1) average complexity). Dictionaries provide faster lookups than lists due to hash-based key access.
15. Can you modify the elements of a tuple? Explain why or why not?
-> No, you cannot modify the elements of a tuple because tuples are immutable in Python. Once created, their elements cannot be changed, added, or removed. This is because Tuples use less memory and allow faster access. Moreover, tuples prevent accidental modifications, making them safer for critical data.
16. What is a nested dictionary, and give an example of its use case?
-> A nested dictionary is a dictionary that contains another dictionary as a value, allowing hierarchical data storage.

Nested dictionaries are useful for organizing complex data, like storing multiple students' details.

students = {
    "101": {"name": "Disha", "age": 21, "course": "CSE"},
    "102": {"name": "Shivangi", "age": 22, "course": "IT"},
}

print(students["101"]["name"])  # Output: Disha
17. Describe the time complexity of accessing elements in a dictionary?
-> Accessing elements in a dictionary has an average time complexity of O(1) because dictionaries use a hash table to store key-value pairs. This allows direct lookups without searching through all elements.

student = {"name": "Disha", "age": 21}

print(student["name"])  # O(1) lookup, Output: Disha
18. In what situations are lists preferred over dictionaries?
-> Lists are preferred over dictionaries when order matters, as they maintain element sequence. They are also better for sequential access, where elements are retrieved by index rather than a key. Additionally, lists consume less memory since they store only values, unlike dictionaries that store key-value pairs. Overall, lists are ideal for ordered, simple data storage, whereas dictionaries are better for fast lookups and key-based access.

19. Why are dictionaries considered unordered, and how does that affect data retrieval?
->Dictionaries were considered unordered before Python 3.7 because they stored key-value pairs in a hash table, where insertion order was not guaranteed. However, from Python 3.7+, dictionaries preserve insertion order while still providing O(1) average-time complexity for lookups.
20. Explain the difference between a list and a dictionary in terms of data retrieval.
-> The main difference between a list and a dictionary in terms of data retrieval is how elements are accessed and their efficiency.

 Lists use index-based retrieval, meaning elements are accessed by their position (O(1) complexity for direct access). However, searching for a specific value requires O(n) complexity.

 Dictionaries use key-based retrieval, allowing fast lookups with O(1) average-time complexity due to hashing. Searching for a value (not a key) still takes O(n).
